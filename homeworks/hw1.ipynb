## Problem 1: Single-Variable Linear Regression

In this problem, you will build and train a single-variable Linear Regression model using PyTorch to predict housing prices based on house size.

The relationship between house size $x$ (in thousands of sq ft) and house price $y$ (in hundreds of thousands of dollars) is modeled as:
$$y = w \cdot x + b + \epsilon$$

where $w$ is the weight (slope), $b$ is the bias (intercept), and $\epsilon \sim \mathcal{N}(0, \sigma^2)$ represents noise.

---
### Your Tasks:
1. Complete the `LinearRegressionModel` class using PyTorch's `nn.Module`.
2. Select the appropriate loss function and optimizer.
3. Write the training loop (forward pass, loss calculation, backward pass, optimizer step).

```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

# Set random seed for reproducibility
torch.manual_seed(42)

# Generate synthetic dataset: 100 housing data points
# True parameters: w = 2.5, b = 1.0
X = torch.rand(100, 1) * 5.0  # House sizes between 0 and 5,000 sq ft
true_w, true_b = 2.5, 1.0
noise = torch.randn(100, 1) * 0.8
y = true_w * X + true_b + noise

print(f"Data generated successfully!")
print(f"X shape: {X.shape} | y shape: {y.shape}")
```

```python
# =====================================================================
# TASK 1: DEFINE THE MODEL
# =====================================================================
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super(LinearRegressionModel, self).__init__()
        # TODO: Define a linear layer with 1 input feature and 1 output feature
        ### YOUR CODE HERE ###
        self.linear = nn.Linear(1, 1)
        ### END YOUR CODE ###

    def forward(self, x):
        # TODO: Define the forward pass
        ### YOUR CODE HERE ###
        return self.linear(x)
        ### END YOUR CODE ###


# Instantiate model
model = LinearRegressionModel()

# =====================================================================
# TASK 2: LOSS FUNCTION & OPTIMIZER
# =====================================================================
# TODO: Choose Mean Squared Error Loss and SGD Optimizer (learning rate = 0.05)
### YOUR CODE HERE ###
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.05)
### END YOUR CODE ###


# =====================================================================
# TASK 3: TRAINING LOOP
# =====================================================================
epochs = 200

for epoch in range(epochs):
    # Set model to training mode
    model.train()

    # TODO: Complete the 5 training loop steps
    ### YOUR CODE HERE ###
    # 1. Forward pass
    predictions = model(X)

    # 2. Compute loss
    loss = criterion(predictions, y)

    # 3. Clear gradients from previous step
    optimizer.zero_grad()

    # 4. Backward pass (compute gradients)
    loss.backward()

    # 5. Update weights
    optimizer.step()
    ### END YOUR CODE ###

    # Print progress every 40 epochs
    if (epoch + 1) % 40 == 0:
        print(f"Epoch [{epoch+1}/{epochs}] - Loss: {loss.item():.4f}")
```

```python
# Plot fitted line vs original data
plt.figure(figsize=(8, 5))
plt.scatter(X.numpy(), y.numpy(), color='blue', label='Ground Truth Data', alpha=0.6)

# Generate predictions for line plot
with torch.no_grad():
    y_pred = model(X)

plt.plot(X.numpy(), y_pred.numpy(), color='red', linewidth=2, label='Fitted Line (PyTorch)')
plt.xlabel('House Size (1,000s sq ft)')
plt.ylabel('Price ($100k)')
plt.title('Linear Regression Model Fit')
plt.legend()
plt.grid(True)
plt.show()
```
